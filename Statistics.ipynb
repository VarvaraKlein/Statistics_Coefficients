```python
import numpy as np

# Фиксируем генератор случайных чисел для воспроизводимости
rng = np.random.default_rng(42)

# Количество виртуальных "экспериментов" (чем больше, тем выше точность)
simulations = 1_000_000

print(f"{'n':<4} | {'Наш расчет (P=0.95)':<20} | {'Таблица из методички'}")
print("-" * 50)

# Эталонные табличные значения (двусторонний Q-тест) для сравнения
book_table = {3: 0.941, 4: 0.765, 5: 0.642, 6: 0.560, 7: 0.507}

for n, book_val in book_table.items():
    # 1. Генерируем 1 млн выборок размером n из стандартного нормального распределения
    data = rng.normal(loc=0, scale=1, size=(simulations, n))

    # 2. Сортируем каждую выборку по возрастанию (вдоль каждой строки)
    data.sort(axis=1)

    # 3. Считаем размах (max - min) для каждой выборки
    data_range = data[:, -1] - data[:, 0]

    # 4. Считаем Q для минимального и максимального элементов и берем больший из них
    q_min = (data[:, 1] - data[:, 0]) / data_range
    q_max = (data[:, -1] - data[:, -2]) / data_range
    q_stat = np.maximum(q_min, q_max)

    # 5. Находим 95-й перцентиль — это и есть критическое значение для P = 0.95!
    q_crit = np.percentile(q_stat, 95)

    print(f"{n:<4} | {q_crit:<20.3f} | {book_val:.3f}")
```

```python
import matplotlib.pyplot as plt
import numpy as np

rng = np.random.default_rng(42)
n = 5  # размер выборки (для примера n = 5, эталон = 0.710)
exact_val = 0.710  # точное табличное значение

# Генерируем один общий массив на 100 000 симуляций, чтобы не перегружать ОЗУ
max_sim = 100_000
data = rng.normal(0, 1, size=(max_sim, n))
data.sort(axis=1)

data_range = data[:, -1] - data[:, 0]
q_min = (data[:, 1] - data[:, 0]) / data_range
q_max = (data[:, -1] - data[:, -2]) / data_range
q_all = np.maximum(q_min, q_max)

# Выбираем 60 точек от 10 до 500 000 в логарифмическом масштабе
sim_counts = np.unique(np.geomspace(10, max_sim, num=60, dtype=int))
q_estimates = [np.percentile(q_all[:k], 95) for k in sim_counts]

# Строим график
plt.figure(figsize=(8, 4.5), dpi=120)
plt.plot(
    sim_counts, q_estimates, 'o-', markersize=4, label='Расчет Монте-Карло'
)
plt.axhline(
    exact_val,
    color='red',
    linestyle='--',
    label=f'Точное значение ({exact_val})',
)

plt.xscale('log')  # логарифмическая шкала по X, чтобы видеть и 100, и 100 000
plt.title(f'Выход расчета Q-критерия на плато (для n = {n}, P = 0.95)')
plt.xlabel('Число симуляций (логарифмическая шкала)')
plt.ylabel('Критическое значение Q')
plt.grid(True, which='both', linestyle='--', alpha=0.5)
plt.legend()
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
from scipy import stats

# 1. Генерируем тренировочные "экспериментальные" данные (5 температур по 5 опытов)
rng = np.random.default_rng(42)
temps = [100, 150, 200, 250, 300]
true_means = [12.0, 24.5, 48.0, 82.0, 115.0]
n_runs = 10  # число повторных опытов при каждой температуре (попробуйте поменять на 3 или 10!)

rows = []
for t, m in zip(temps, true_means):
    # Добавляем случайный разброс (шум прибора ~7% от значения)
    measurements = rng.normal(loc=m, scale=m * 0.07, size=n_runs)
    for val in measurements:
        rows.append({'temperature': t, 'activity': val})

df_train = pd.DataFrame(rows)


# 2. Функция расчета среднего и доверительного интервала
def get_ci_table(data, conf):
    def calc(g):
        n = len(g)
        mean = g.mean()
        ci = (g.std(ddof=1) / np.sqrt(n)) * stats.t.ppf(
            (1 + conf) / 2, df=n - 1
        )
        return pd.Series({'mean': mean, 'ci': ci})

    return data.groupby('temperature')['activity'].apply(calc).unstack()


res_95 = get_ci_table(df_train, conf=0.95)
res_99 = get_ci_table(df_train, conf=0.99)

# 3. Рисуем график сравнения
plt.figure(figsize=(8, 5), dpi=120)

# Широкие внешние усы (P = 0.99)
plt.errorbar(
    res_99.index,
    res_99['mean'],
    yerr=res_99['ci'],
    fmt='none',
    ecolor='#ff7f0e',
    elinewidth=3.5,
    capsize=7,
    alpha=0.55,
    label='ДИ для P = 0.99 (шире)',
)

# Узкие внутренние усы и кривая (P = 0.95)
plt.errorbar(
    res_95.index,
    res_95['mean'],
    yerr=res_95['ci'],
    fmt='o-',
    color='#1f77b4',
    ecolor='#1f77b4',
    elinewidth=1.5,
    capsize=4,
    markersize=6,
    label='Среднее ± ДИ для P = 0.95 (стандарт)',
)

plt.title(
    f'Тренировочный график: сравнение ДИ (число опытов n = {n_runs})',
    fontsize=12,
)
plt.xlabel('Температура, °C', fontsize=11)
plt.ylabel('Каталитическая активность', fontsize=11)
plt.grid(True, linestyle='--', alpha=0.5)
plt.legend()
plt.tight_layout()
plt.show()
```
